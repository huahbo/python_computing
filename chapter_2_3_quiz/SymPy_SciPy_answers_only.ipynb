## 题 7 参考答案代码

```python
try:
    from scipy.optimize import brent
    xmin_brent = brent(lambda t: (t-1)**2, brack=(0,2))
except Exception:
    from scipy.optimize import minimize_scalar
    res = minimize_scalar(lambda t: (t-1)**2, bounds=(0,2), method='bounded')
    xmin_brent = float(res.x)
print(xmin_brent)
```

## 题 8 参考答案代码

```python
import numpy as np
from scipy.optimize import linear_sum_assignment
cost_matrix = np.array([[4,1,3],[2,0,5],[3,2,2]])
row_ind, col_ind = linear_sum_assignment(cost_matrix)
min_cost_val = int(cost_matrix[row_ind, col_ind].sum())
print(min_cost_val)
```

## 题 14 参考答案代码

```python
import numpy as np
from scipy.integrate import odeint
def f_sys(y, t):
    y1, y2 = y
    return [y2, -y1]
t = np.linspace(0, 2*np.pi, 100)
y0 = [1, 0]
y_sol = odeint(f_sys, y0, t)[:,0]
print(y_sol[:5])
```

## 题 15 参考答案代码

```python
import numpy as np
from scipy.optimize import curve_fit
np.random.seed(0)
def model(x, a, b, c): return a*x**2 + b*x + c
xdata = np.linspace(-5,5,100)
ydata = model(xdata, 2, 3, 1) + np.random.normal(0, 0.2, xdata.shape)
params, pcov = curve_fit(model, xdata, ydata, p0=[1,1,1])
print(params)
```

## 题 18 参考答案代码

```python
import numpy as np
from scipy.fft import rfft, rfftfreq
fs = 100
T = 1.0
t = np.linspace(0, T, fs, endpoint=False)
sig = np.sin(2*np.pi*5*t)
Y = rfft(sig)
freqs = rfftfreq(len(t), 1/fs)
peak_freq = float(freqs[np.argmax(np.abs(Y))])
print(peak_freq)
```

## 题 21 参考答案代码

```python
from scipy.optimize import minimize
import numpy as np
def rosen(vars, a=1, b=100):
    x,y = vars
    return (a-x)**2 + b*(y-x**2)**2
res = minimize(rosen, [-1.2,1.0], method='BFGS')
rosen_x = list(res.x)
print(rosen_x)
```

## 题 24 参考答案代码

```python
from scipy.integrate import solve_ivp
import numpy as np
def lorenz(t, state, sigma=10., rho=28., beta=8/3):
    x,y,z = state
    return [sigma*(y-x), x*(rho-z)-y, x*y - beta*z]
sol_lorenz = solve_ivp(lambda t, y: lorenz(t,y), [0,2], [0.,1.,0.], t_eval=np.linspace(0,2,200))
print(sol_lorenz.y.shape)
```

## 题 26 参考答案代码

```python
import numpy as np
from scipy.optimize import curve_fit
np.random.seed(0)
ts = np.linspace(0,1,200)
true_A = 2.0; true_f = 5.0; true_phi = 0.3
ys = true_A * np.sin(2*np.pi*true_f*ts + true_phi) + np.random.normal(0, 0.1, len(ts))
def sin_model(t, A, f, phi): return A * np.sin(2*np.pi*f*t + phi)
popt, pcov = curve_fit(sin_model, ts, ys, p0=[1.0,5.0,0.0], maxfev=20000)
est_f = float(popt[1])
print(est_f)
```

## 题 27 参考答案代码

```python
import sympy as sp
x,y,lam = sp.symbols('x y lam')
fx = x**2 + y**2
g = x + y - 1
L = fx + lam*g
sol = sp.solve([sp.diff(L,v) for v in (x,y,lam)], [x,y,lam], dict=True)
opt_xy = sol
first = sol[0]
opt_x = sp.N(first[x])
opt_y = sp.N(first[y])
print(opt_x, opt_y)
```

## 题 28 参考答案代码

```python
from scipy.optimize import linprog
import numpy as np
c = np.array([1.0,1.0])
A_ub = np.array([[1.0,2.0]])
b_ub = np.array([4.0])
res_lp = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=[(0,None),(0,None)], method='highs')
lp_x = list(res_lp.x)
print(lp_x)
```

## 题 29 参考答案代码

```python
import sympy as sp, numpy as np
x = sp.symbols('x')
expr = sp.sin(x)**2
deriv = sp.diff(expr, x)
fnum = sp.lambdify(x, expr, 'numpy')
deriv_num = sp.lambdify(x, deriv, 'numpy')
xx = 0.7
h = 1e-6
fd = (fnum(xx+h) - fnum(xx-h)) / (2*h)
err = abs(fd - deriv_num(xx))
print(err)
```

## 题 30 参考答案代码

```python
import sympy as sp
r,K = sp.symbols('r K', positive=True)
t = sp.symbols('t')
x_fun = sp.Function('x')
sol_logistic = sp.dsolve(sp.Eq(sp.diff(x_fun(t), t), r*x_fun(t)*(1 - x_fun(t)/K)), ics={x_fun(0): sp.Rational(1,10)})
print(sol_logistic)
```
