## 题 03 解答（容易）— DFS 与 BFS 遍历顺序

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import networkx as nx

G = nx.Graph()
G.add_edges_from([(1,2),(1,3),(2,3),(3,4),(4,5)])
dfs_nodes = list(nx.dfs_preorder_nodes(G, source=1))
seen = set()
bfs_nodes = []
for u, v in nx.bfs_edges(G, source=1):
    if u not in seen:
        seen.add(u); bfs_nodes.append(u)
    if v not in seen:
        seen.add(v); bfs_nodes.append(v)
print(dfs_nodes)
print(bfs_nodes)
```

## 题 06 解答（中等）— 带权最短路径（Dijkstra）

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import networkx as nx

DG = nx.DiGraph()
DG.add_edge('A','B',weight=3)
DG.add_edge('A','C',weight=2)
DG.add_edge('B','C',weight=1)
DG.add_edge('B','D',weight=4)
DG.add_edge('C','D',weight=2)
DG.add_edge('C','E',weight=1)
DG.add_edge('D','E',weight=3)
sp = nx.dijkstra_path(DG, 'A', 'E', weight='weight')
sp_len = nx.dijkstra_path_length(DG, 'A', 'E', weight='weight')
print(sp, sp_len)
```

## 题 07 解答（中等）— 最小生成树（MST）与总权重

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import networkx as nx

G = nx.Graph()
edges = [
    ('A','B',10),('A','C',6),('A','D',5),('B','C',5),('B','D',15),
    ('B','E',4),('C','D',15),('C','F',12),('D','E',20),('E','F',9)
]
G.add_weighted_edges_from(edges)
T = nx.minimum_spanning_tree(G)
total_w = sum(w for *_, w in T.edges(data='weight'))
print(sorted(T.edges(data=True)))
print(total_w)
```
