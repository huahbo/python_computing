### 题 20

综合题：实现一个小功能（编程）

<details><summary>提示（点击展开）</summary>

提示：使用 cumsum，注意前缀和技巧。

<details><summary>答案（代码，点击展开可查看）</summary>

```python
import numpy as np
def moving_average(arr, k):
    arr = np.asarray(arr, dtype=float)
    if k <= 0 or k > arr.size:
        raise ValueError("k must be between 1 and len(arr)")
    c = np.cumsum(arr)
    out = (c[k-1:] - np.concatenate(([0.0], c[:-k]))) / k
    return out
_q20 = moving_average(np.arange(10), 3)
print(_q20)

```

</details>


````python
import numpy as np
def moving_average(arr, k):
    arr = np.asarray(arr, dtype=float)
    if k <= 0 or k > arr.size:
        raise ValueError("k must be between 1 and len(arr)")
    c = np.cumsum(arr)
    out = (c[k-1:] - np.concatenate(([0.0], c[:-k]))) / k
    return out
_q20 = moving_average(np.arange(10), 3)
print(_q20)
````
