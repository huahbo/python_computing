## 题 1 解答（容易）— 用 OLS 进行一元线性回归并输出 R²

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np
import statsmodels.api as sm

rng = np.random.default_rng(0)
x = np.linspace(0, 10, 50)
y = 2*x + 1 + rng.normal(0, 1, size=len(x))

X = sm.add_constant(x.reshape(-1,1))
model = sm.OLS(y, X).fit()
print(model.params)
print(model.rsquared)
```

## 题 3 解答（容易）— 单因素方差分析（One-Way ANOVA）

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np, pandas as pd, statsmodels.api as sm
from statsmodels.formula.api import ols

rng = np.random.default_rng(1)
n = 12
df = pd.DataFrame({
    'group': ['A']*n + ['B']*n + ['C']*n,
    'y': np.r_[rng.normal(0,1,n), rng.normal(1,1,n), rng.normal(2,1,n)]
})
model = ols('y ~ C(group)', data=df).fit()
anova_res = sm.stats.anova_lm(model, typ=1)
print(anova_res)
```

## 题 4 解答（容易）— GLM（二项 Logit）拟合与概率预测

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np, statsmodels.api as sm

rng = np.random.default_rng(0)
X_raw = rng.uniform(0,1,(100,2))
lin = 0.5 + 1.2*X_raw[:,0] - 0.8*X_raw[:,1]
p = 1/(1+np.exp(-lin))
Y = rng.binomial(1, p)

X = sm.add_constant(X_raw)
model = sm.GLM(Y, X, family=sm.families.Binomial()).fit()
X_new = sm.add_constant(np.array([[0.2,0.3],[0.8,0.1]]))
print(model.predict(X_new))
```

## 题 6 解答（中等）— 时间序列分解（seasonal_decompose）

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

rng = np.random.default_rng(0)
idx = pd.date_range('2022-01-01', periods=180, freq='D')
trend = np.linspace(0, 5, len(idx))
season = 2*np.sin(2*np.pi*np.arange(len(idx))/7)
noise = rng.normal(0, 0.5, len(idx))
y = 10 + trend + season + noise
s = pd.Series(y, index=idx)
result = seasonal_decompose(s, model='additive', period=7)
result.plot()
plt.show()
```

## 题 7 解答（中等）— ADF 单位根检验

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np
from statsmodels.tsa.stattools import adfuller

rng = np.random.default_rng(0)
n = 300
y = np.zeros(n)
for t in range(1, n):
    y[t] = 0.7*y[t-1] + rng.normal(0, 1)
stat, pvalue, *rest = adfuller(y)
print(pvalue)
```

## 题 8 解答（稍难）— ARIMA(1,1,1) 拟合与 5 步预测

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np, pandas as pd
from statsmodels.tsa.arima.model import ARIMA

rng = np.random.default_rng(0)
n = 200
steps = rng.normal(0, 1, n).cumsum() + np.linspace(0, 10, n)
s = pd.Series(steps)
model = ARIMA(s, order=(1,1,1)).fit()
fc = model.forecast(steps=5)
print(fc)
```

## 题 9 解答（稍难）— 加权最小二乘（WLS）处理异方差

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np, statsmodels.api as sm

rng = np.random.default_rng(1)
x = np.linspace(0.1, 10, 200)
eps = rng.normal(0, x)
y = 3*x + eps

X = sm.add_constant(x.reshape(-1,1))
ols_res = sm.OLS(y, X).fit()
wls_res = sm.WLS(y, X, weights=1/np.square(x)).fit()
print(ols_res.params, wls_res.params)
```

## 题 10 解答（稍难）— Logit 阈值选择对分类指标的影响

**思路**：参见题面提示，按库函数正确调用并注意数据预处理/参数设置。下方给出参考实现。

```python
import numpy as np, statsmodels.api as sm
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(0)
X_raw = rng.uniform(0,1,(300,2))
lin = 0.2 + 2.0*X_raw[:,0] - 1.5*X_raw[:,1]
p = 1/(1+np.exp(-lin))
Y = rng.binomial(1, p)
X_train, X_test, y_train, y_test = train_test_split(X_raw, Y, test_size=0.3, random_state=42)

X_tr = sm.add_constant(X_train); X_te = sm.add_constant(X_test)
model = sm.GLM(y_train, X_tr, family=sm.families.Binomial()).fit()
proba = model.predict(X_te)
for thr in [0.3,0.5,0.7]:
    y_pred = (proba>=thr).astype(int)
    print(thr, accuracy_score(y_test, y_pred), f1_score(y_test, y_pred))
```
